## ۱۱.۱ مقدمه

## ۱۱.۳ استثناهای رایج در پایتون

| نام استثنا | زمان رخ دادن | مثال |
|---|---|---|
| `ZeroDivisionError` | تقسیم بر صفر | `10 / 0` |
| `TypeError` | عملیات روی نوع داده اشتباه | `"hello" + 5` |
| `ValueError` | مقدار نامعتبر برای تابع | `int("abc")` |
| `NameError` | استفاده از متغیر تعریف‌نشده | `print(x)` بدون تعریف `x` |
| `IndexError` | دسترسی به ایندکس خارج از محدوده | `lst[10]` در لیست ۳ عنصری |
| `KeyError` | کلید ناموجود در دیکشنری | `d["missing_key"]` |
| `FileNotFoundError` | فایل وجود ندارد | `open("no_file.txt")` |
| `AttributeError` | صفت ناموجود روی شیء | `"hello".push("x")` |
| `ImportError` | ماژول یافت نشد | `import nonexistent_module` |
| `OverflowError` | عدد خیلی بزرگ برای محاسبه | `math.exp(1000)` |

## ۱۱.۴ ساختار `try` و `except`

ساده‌ترین روش مدیریت استثنا:
</div>


```
try:
    # کدی که ممکن است خطا بدهد
    number = int(input("یک عدد وارد کنید: "))
    result = 100 / number
    print(f"نتیجه: {result}")
except ZeroDivisionError:
    print("خطا: نمی‌توان بر صفر تقسیم کرد.")
except ValueError:
    print("خطا: لطفاً یک عدد صحیح وارد کنید.")
```

اگر کاربر `abc` وارد کند:

خطا: لطفاً یک عدد صحیح وارد کنید.

## ۱۱.۵ گرفتن چند استثنا

### روش ۱ — چند بلوک `except` جداگانه
</div>


```
def safe_divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        print("خطا: تقسیم بر صفر.")
        return None
    except TypeError:
        print("خطا: ورودی‌ها باید عدد باشند.")
        return None

print(safe_divide(10, 2))   # 5.0
print(safe_divide(10, 0))   # خطا: تقسیم بر صفر.
print(safe_divide(10, "a")) # خطا: ورودی‌ها باید عدد باشند.
```

<div dir="rtl" style="text-align:right;">

### روش ۲ — گرفتن چند استثنا در یک `except`

```
try:
    value = int(input("عدد: "))
    result = 100 / value
except (ValueError, ZeroDivisionError) as e:
    print(f"خطا رخ داد: {e}")
```

## ۱۱.۶ بلوک `else`

بلوک `else` فقط وقتی اجرا می‌شود که **هیچ استثنایی** رخ نداده باشد.
</div>


```
try:
    number = int(input("عدد: "))
    result = 100 / number
except ValueError:
    print("ورودی معتبر نیست.")
except ZeroDivisionError:
    print("تقسیم بر صفر.")
else:
    # فقط اگر try بدون خطا تمام شود
    print(f"عملیات موفق بود. نتیجه: {result}")
```

## ۱۱.۷ بلوک `finally`

بلوک `finally` **همیشه** اجرا می‌شود، چه خطا رخ داده باشد چه نه. برای آزادسازی منابع ایده‌آل است.
</div>


```
file = None
try:
    file = open("data.txt", "r")
    content = file.read()
    print(content)
except FileNotFoundError:
    print("فایل یافت نشد.")
finally:
    # این بلوک همیشه اجرا می‌شود
    if file:
        file.close()
        print("فایل بسته شد.")
```

## ۱۱.۸ ایجاد و پرتاب استثنا با `raise`

گاهی می‌خواهید خودتان یک استثنا ایجاد کنید تا به کد فراخواننده اطلاع دهید که مشکلی وجود دارد.
</div>


```
def set_age(age):
    if not isinstance(age, int):
        raise TypeError("سن باید یک عدد صحیح باشد.")
    if age < 0 or age > 150:
        raise ValueError(f"سن {age} معتبر نیست. باید بین 0 و 150 باشد.")
    return age


try:
    print(set_age(25))    # 25
    print(set_age(-5))    # خطا
except ValueError as e:
    print(f"ValueError: {e}")
except TypeError as e:
    print(f"TypeError: {e}")
```

<div dir="rtl" style="text-align:right;">

خروجی:

25
ValueError: سن -5 معتبر نیست. باید بین 0 و 150 باشد.

## ۱۱.۹ استثناهای سفارشی (Custom Exceptions)

می‌توانید کلاس‌های استثنای خودتان را بسازید تا خطاهای مخصوص برنامه‌تان را مدیریت کنید.
</div>


<div dir="rtl" style="text-align:right;">

# تعریف استثناهای سفارشی
</div>


```
class AppError(Exception):
    """کلاس پایه برای خطاهای این برنامه"""
    pass

class InvalidCoordinateError(AppError):
    """خطا برای مختصات نامعتبر"""
    def __init__(self, lat, lon, message="مختصات نامعتبر است"):
        self.lat = lat
        self.lon = lon
        self.message = message
        super().__init__(self.message)

    def __str__(self):
        return f"{self.message}: lat={self.lat}, lon={self.lon}"

class DataOutOfRangeError(AppError):
    """خطا برای داده خارج از محدوده"""
    pass
```

<div dir="rtl" style="text-align:right;">

# استفاده از استثناهای سفارشی
</div>


```
def validate_coordinate(lat, lon):
    if not (-90 <= lat <= 90):
        raise InvalidCoordinateError(lat, lon, "عرض جغرافیایی باید بین -90 و 90 باشد")
    if not (-180 <= lon <= 180):
        raise InvalidCoordinateError(lat, lon, "طول جغرافیایی باید بین -180 و 180 باشد")
    return True


try:
    validate_coordinate(200, 51)
except InvalidCoordinateError as e:
    print(f"خطای مختصات: {e}")
except AppError as e:
    print(f"خطای برنامه: {e}")
```

<div dir="rtl" style="text-align:right;">

خروجی:

خطای مختصات: عرض جغرافیایی باید بین -90 و 90 باشد: lat=200, lon=51

## ۱۱.۱۰ زنجیره استثناها (Exception Chaining)

وقتی در داخل یک `except` استثنای جدیدی پرتاب می‌کنید، می‌توانید استثنای اصلی را حفظ کنید.
</div>


```
def read_config(filename):
    try:
        with open(filename, "r") as f:
            return f.read()
    except FileNotFoundError as e:
        raise RuntimeError(f"فایل تنظیمات '{filename}' یافت نشد.") from e


try:
    config = read_config("config.txt")
except RuntimeError as e:
    print(f"خطا: {e}")
    print(f"علت اصلی: {e.__cause__}")
```

## ۱۱.۱۱ مثال کاربردی: سیستم ورودی امن
</div>


```
def get_valid_input(prompt, data_type, min_val=None, max_val=None):
    """
    ورودی معتبر از کاربر می‌گیرد.
    prompt   : پیام نمایشی
    data_type: نوع داده مورد انتظار (int, float, str)
    min_val  : حداقل مقدار مجاز
    max_val  : حداکثر مقدار مجاز
    """
    while True:
        try:
            raw = input(prompt)
            value = data_type(raw)

            if min_val is not None and value < min_val:
                raise ValueError(f"مقدار باید حداقل {min_val} باشد.")
            if max_val is not None and value > max_val:
                raise ValueError(f"مقدار باید حداکثر {max_val} باشد.")

            return value

        except ValueError as e:
            print(f"  ورودی نامعتبر: {e}. دوباره تلاش کنید.")
        except KeyboardInterrupt:
            print("\nعملیات توسط کاربر لغو شد.")
        return None
```

<div dir="rtl" style="text-align:right;">

# استفاده
</div>


```
age   = get_valid_input("سن خود را وارد کنید: ", int, min_val=0, max_val=120)
score = get_valid_input("نمره (0-100): ", float, min_val=0.0, max_val=100.0)

if age is not None and score is not None:
    print(f"سن: {age}, نمره: {score}")
```

## ۱۱.۱۲ مثال کاربردی: خواندن امن فایل CSV
</div>


```
import csv

class CSVReadError(Exception):
    pass


def read_station_data(filepath):
    stations = []
    try:
        with open(filepath, "r", encoding="utf-8") as f:
            reader = csv.DictReader(f)

            required_columns = {"City", "Latitude", "Longitude", "Elevation"}
            if not required_columns.issubset(reader.fieldnames):
                missing = required_columns - set(reader.fieldnames)
                raise CSVReadError(f"ستون‌های ناموجود: {missing}")

            for row_num, row in enumerate(reader, start=2):
                try:
                    station = {
                        "city":      row["City"],
                        "latitude":  float(row["Latitude"]),
                        "longitude": float(row["Longitude"]),
                        "elevation": int(row["Elevation"]),
                    }
                    stations.append(station)
                except ValueError as e:
                    print(f"  هشدار: ردیف {row_num} نادیده گرفته شد — {e}")

    except FileNotFoundError:
        raise CSVReadError(f"فایل '{filepath}' یافت نشد.")
    except Exception as e:
        raise CSVReadError(f"خطا در خواندن فایل: {e}") from e

    return stations


try:
    data = read_station_data("stations.csv")
    print(f"{len(data)} ایستگاه بارگذاری شد.")
    for s in data:
        print(f"  {s['city']}: {s['elevation']} m")
except CSVReadError as e:
    print(f"خطای CSV: {e}")
```

## ۱۱.۱۳ نکات مهم

> **نکته ۱ — خاص‌ترین استثنا را اول بنویسید:** پایتون بلوک‌های `except` را از بالا به پایین بررسی می‌کند. اگر `Exception` را اول بنویسید، بقیه هرگز اجرا نمی‌شوند.

# اشتباه
</div>


```
try:
    result = 10 / 0
except Exception:
    print("خطای کلی")
except ZeroDivisionError:   # هرگز به اینجا نمی‌رسد
    print("تقسیم بر صفر")
```

<div dir="rtl" style="text-align:right;">

# درست
</div>


```
try:
    result = 10 / 0
except ZeroDivisionError:
    print("تقسیم بر صفر")
except Exception:
    print("خطای کلی")
```

<div dir="rtl" style="text-align:right;">

> **نکته ۲ — از `except:` خالی پرهیز کنید:** این حالت حتی `KeyboardInterrupt` و `SystemExit` را هم می‌گیرد و می‌تواند برنامه را غیرقابل توقف کند.

> **نکته ۳ — پیام خطا را لاگ کنید:** در برنامه‌های واقعی، خطاها را در فایل لاگ ذخیره کنید تا بتوانید بعداً بررسی کنید.
</div>


```
import logging

logging.basicConfig(
    filename="app.log",
    level=logging.ERROR,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

try:
    result = 10 / 0
except ZeroDivisionError as e:
    logging.error(f"خطای تقسیم: {e}")
    print("خطایی رخ داد. جزئیات در فایل لاگ ثبت شد.")
```

<div dir="rtl" style="text-align:right;">

> **نکته ۴ — `finally` برای آزادسازی منابع:** اتصال به پایگاه داده، فایل‌های باز، و اتصالات شبکه را همیشه در `finally` ببندید.

## ۱۱.۱۴ تمرین‌های عملی

### تمرین ۱ — ماشین حساب امن
یک ماشین حساب بنویسید که چهار عمل اصلی را انجام دهد و تمام خطاهای ممکن (تقسیم بر صفر، ورودی غیرعددی، عملگر نامعتبر) را مدیریت کند.

### تمرین ۲ — خواندن امن فایل
برنامه‌ای بنویسید که نام فایل را از کاربر بگیرد، آن را بخواند و در صورت نبود فایل، از کاربر بخواهد نام دیگری وارد کند (حداکثر ۳ بار تلاش).

### تمرین ۳ — اعتبارسنجی داده
یک کلاس `Temperature` بسازید که دما را در سلسیوس ذخیره کند. اگر دما کمتر از `273.15-` (صفر مطلق) باشد، استثنای سفارشی `InvalidTemperatureError` پرتاب کند.

### تمرین ۴ — لاگ‌گذاری خطاها
برنامه‌ای بنویسید که یک لیست از اعداد را از فایل بخواند، میانگین آن‌ها را حساب کند و هر خطا (فایل ناموجود، مقدار غیرعددی، لیست خالی) را در فایل `errors.log` ثبت کند.

### تمرین ۵ — زنجیره استثنا
یک تابع `load_config(path)` بنویسید که فایل JSON را بخواند و تجزیه کند. خطاهای `FileNotFoundError` و `json.JSONDecodeError` را به یک `ConfigError` سفارشی تبدیل کنید و علت اصلی را حفظ کنید.
