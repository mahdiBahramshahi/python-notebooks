# فصل ۳: عملگرها و عبارات در پایتون

## ۳.۱ مقدمه

در فصل قبل با متغیرها و انواع داده آشنا شدیم. حالا می‌خواهیم با این داده‌ها **کار کنیم** — محاسبه کنیم، مقایسه کنیم، و شرط بگذاریم. در این فصل با **عملگرها** (Operators) در پایتون آشنا می‌شویم.

در این فصل یاد می‌گیریم:
- عملگرهای حسابی برای محاسبات عددی
- عملگرهای مقایسه‌ای برای مقایسه مقادیر
- عملگرهای منطقی برای ترکیب شرط‌ها
- عملگرهای انتساب ترکیبی برای کوتاه‌نویسی
- اولویت عملگرها برای اجرای صحیح محاسبات

---

## ۳.۲ عملگرهای حسابی (Arithmetic Operators)

این عملگرها برای انجام عملیات ریاضی پایه استفاده می‌شوند.

| عملگر | عملکرد | مثال | نتیجه |
|---|---|---|---|
| `+` | جمع | `5 + 3` | `8` |
| `-` | تفریق | `5 - 3` | `2` |
| `*` | ضرب | `5 * 3` | `15` |
| `/` | تقسیم (نتیجه float) | `5 / 2` | `2.5` |
| `//` | تقسیم صحیح | `5 // 2` | `2` |
| `%` | باقیمانده تقسیم | `5 % 2` | `1` |
| `**` | توان | `2 ** 8` | `256` |
</div>


````
a = 10
b = 3

print(a + b)    # 13
print(a - b)    # 7
print(a * b)    # 30
print(a / b)    # 3.3333...
print(a // b)   # 3
print(a % b)    # 1
print(a ** b)   # 1000
````

<div dir="rtl" style="text-align:right;">

> **نکته مهم:** عملگر `/` همیشه یک عدد اعشاری (float) برمی‌گرداند، حتی اگر نتیجه تقسیم عدد صحیح باشد. برای دریافت عدد صحیح از تقسیم، از `//` استفاده کنید.

---

## ۳.۳ مثال کاربردی: محاسبات زمین‌شناختی

### محاسبه NDVI (شاخص نرمال‌شده تفاوت پوشش گیاهی)

فرمول NDVI:
</div>


<div dir="rtl" style="text-align:right;">

NDVI = (NIR - Red) / (NIR + Red)
</div>


````
nir_reflectance = 0.45  # بازتابندگی باند مادون قرمز نزدیک
red_reflectance = 0.12  # بازتابندگی باند قرمز

ndvi = (nir_reflectance - red_reflectance) / (nir_reflectance + red_reflectance)
print(f"NDVI = {ndvi:.4f}")
````

<div dir="rtl" style="text-align:right;">

خروجی:

NDVI = 0.5789

### محاسبه گرادیان دمایی زمین‌گرمایی

فرمول گرادیان دمایی:
</div>


<div dir="rtl" style="text-align:right;">

G = (T_depth - T_surface) / Depth
</div>


````
surface_temp = 85.0     # دمای سطح (°C)
bottom_temp = 250.0     # دمای کف چاه (°C)
well_depth_km = 5.0     # عمق چاه (km)

geothermal_gradient = (bottom_temp - surface_temp) / well_depth_km
print(f"گرادیان دمایی: {geothermal_gradient:.2f} °C/km")
````

<div dir="rtl" style="text-align:right;">

خروجی:

گرادیان دمایی: 33.00 °C/km

## ۳.۴ عملگرهای مقایسه‌ای (Comparison Operators)

این عملگرها برای مقایسه دو مقدار استفاده می‌شوند و نتیجه آن‌ها همیشه `True` یا `False` است.

| عملگر | معنی | مثال | نتیجه |
|---|---|---|---|
| `==` | مساوی | `5 == 5` | `True` |
| `!=` | نامساوی | `5 != 3` | `True` |
| `>` | بزرگتر | `5 > 3` | `True` |
| `<` | کوچکتر | `5 < 3` | `False` |
| `>=` | بزرگتر یا مساوی | `5 >= 5` | `True` |
| `<=` | کوچکتر یا مساوی | `3 <= 5` | `True` |
</div>


````
ndvi = 0.65

print(ndvi > 0.5)     # True — پوشش گیاهی متراکم
print(ndvi == 1.0)    # False
print(ndvi >= 0.6)    # True
print(ndvi < 0.2)     # False — خاک لخت نیست
````

<div dir="rtl" style="text-align:right;">

> **نکته:** در پایتون، `=` برای انتساب مقدار به متغیر است، در حالی که `==` برای مقایسه تساوی دو مقدار استفاده می‌شود.

---

## ۳.۵ مثال کاربردی: طبقه‌بندی NDVI
</div>


````
ndvi_value = 0.72

is_dense_vegetation = ndvi_value > 0.6
is_sparse_vegetation = ndvi_value > 0.2 and ndvi_value <= 0.6
is_bare_soil = ndvi_value < 0.1

print(f"مقدار NDVI: {ndvi_value}")
print(f"پوشش گیاهی متراکم: {is_dense_vegetation}")
print(f"پوشش گیاهی پراکنده: {is_sparse_vegetation}")
print(f"خاک لخت: {is_bare_soil}")
````

<div dir="rtl" style="text-align:right;">

خروجی:

مقدار NDVI: 0.72
پوشش گیاهی متراکم: True
پوشش گیاهی پراکنده: False
خاک لخت: False

## ۳.۶ عملگرهای منطقی (Logical Operators)

این عملگرها برای ترکیب یا تغییر عبارات بولین به کار می‌روند.

| عملگر | معنی | مثال | نتیجه |
|---|---|---|---|
| `and` | هر دو شرط باید True باشند | `True and True` | `True` |
| `or` | حداقل یکی True باشد | `True or False` | `True` |
| `not` | نتیجه را معکوس می‌کند | `not True` | `False` |

### جدول درستی and

| A | B | A and B |
|---|---|---|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

### جدول درستی or

| A | B | A or B |
|---|---|---|
| True | True | True |
| True | False | True |
| False | True | True |
| False | False | False |
</div>


````
cloud_cover = 5.0
ndvi = 0.65

is_suitable_for_veg_analysis = (cloud_cover < 10) and (ndvi > 0.5)
print(f"مناسب برای تحلیل پوشش گیاهی: {is_suitable_for_veg_analysis}")

needs_further_review = (ndvi < 0.2) or (cloud_cover > 20)
print(f"نیاز به بررسی بیشتر: {needs_further_review}")

is_clear_image = not (cloud_cover > 10)
print(f"تصویر بدون ابر: {is_clear_image}")
````

<div dir="rtl" style="text-align:right;">

خروجی:

مناسب برای تحلیل پوشش گیاهی: True
نیاز به بررسی بیشتر: False
تصویر بدون ابر: True

## ۳.۷ مثال کاربردی: ارزیابی کیفیت داده‌های سنجش از دور
</div>


````
cloud_cover_percentage = 8.5
sun_elevation_angle = 42.0
ndvi_value = 0.58
acquisition_year = 2023

is_low_cloud = cloud_cover_percentage < 10
is_good_sun_angle = sun_elevation_angle > 30
is_valid_ndvi = 0.0 <= ndvi_value <= 1.0
is_recent_data = acquisition_year >= 2020

is_high_quality_scene = is_low_cloud and is_good_sun_angle and is_valid_ndvi and is_recent_data

print(f"پوشش ابر کم: {is_low_cloud}")
print(f"زاویه تابش خورشید مناسب: {is_good_sun_angle}")
print(f"مقدار NDVI معتبر: {is_valid_ndvi}")
print(f"داده اخیر: {is_recent_data}")
print(f"کیفیت کلی صحنه بالا است: {is_high_quality_scene}")
````

## ۳.۸ عملگرهای انتساب ترکیبی (Augmented Assignment Operators)

این عملگرها راهی کوتاه‌تر برای به‌روزرسانی مقدار یک متغیر هستند.

| عملگر | معادل | مثال |
|---|---|---|
| `+=` | `x = x + n` | `depth += 100` |
| `-=` | `x = x - n` | `depth -= 50` |
| `*=` | `x = x * n` | `value *= 2` |
| `/=` | `x = x / n` | `value /= 10` |
| `//=` | `x = x // n` | `count //= 2` |
| `%=` | `x = x % n` | `index %= 5` |
| `**=` | `x = x ** n` | `base **= 2` |
</div>


````
current_depth = 1000

current_depth += 250
print(f"عمق فعلی: {current_depth} متر")

current_depth -= 50
print(f"عمق پس از اصلاح: {current_depth} متر")

current_depth *= 2
print(f"عمق به فوت (تقریبی): {current_depth} فوت")
````

<div dir="rtl" style="text-align:right;">

خروجی:

عمق فعلی: 1250 متر
عمق پس از اصلاح: 1200 متر
عمق به فوت (تقریبی): 2400 فوت

## ۳.۹ عملگرهای عضویت (in و not in)

برای بررسی وجود یک مقدار در یک دنباله استفاده می‌شوند.
</div>


````
rock_type = "Granite"
common_minerals = "quartz, feldspar, mica"

contains_quartz = "quartz" in common_minerals
print(f"حاوی کوارتز است: {contains_quartz}")

contains_calcite = "calcite" in common_minerals
print(f"حاوی کلسیت است: {contains_calcite}")

not_contains_calcite = "calcite" not in common_minerals
print(f"فاقد کلسیت است: {not_contains_calcite}")

is_granite = "Granite" in rock_type
print(f"نوع سنگ گرانیت است: {is_granite}")
````

<div dir="rtl" style="text-align:right;">

خروجی:

حاوی کوارتز است: True
حاوی کلسیت است: False
فاقد کلسیت است: True
نوع سنگ گرانیت است: True

## ۳.۱۰ اولویت عملگرها (Operator Precedence)

پایتون هنگام ارزیابی عبارات، عملگرها را بر اساس **اولویت** اجرا می‌کند.

| اولویت | عملگر |
|---|---|
| ۱ (بالاترین) | `()` پرانتز |
| ۲ | `**` توان |
| ۳ | `+x`, `-x` (عملگرهای یگانی) |
| ۴ | `*`, `/`, `//`, `%` |
| ۵ | `+`, `-` |
| ۶ | `==`, `!=`, `>`, `<`, `>=`, `<=` |
| ۷ | `not` |
| ۸ | `and` |
| ۹ (پایین‌ترین) | `or` |
</div>


<div dir="rtl" style="text-align:right;">

# بدون پرانتز — ضرب زودتر از جمع اجرا می‌شود
</div>


````
result = 2 + 3 * 4
print(result)    # 14  (نه 20)
````

<div dir="rtl" style="text-align:right;">

# با پرانتز — جمع زودتر اجرا می‌شود
</div>


````
result = (2 + 3) * 4
print(result)    # 20
````

<div dir="rtl" style="text-align:right;">

### مثال زمین‌شناختی: اهمیت پرانتز در NDVI
</div>


````
nir = 0.45
red = 0.12
````

<div dir="rtl" style="text-align:right;">

# اشتباه — بدون پرانتز
</div>


````
ndvi_wrong = nir - red / nir + red
print(f"NDVI اشتباه: {ndvi_wrong:.4f}")    # 0.8267
````

<div dir="rtl" style="text-align:right;">

# درست — با پرانتز
</div>


````
ndvi_correct = (nir - red) / (nir + red)
print(f"NDVI درست: {ndvi_correct:.4f}")    # 0.5789
````

<div dir="rtl" style="text-align:right;">

> **توصیه:** در فرمول‌های پیچیده همیشه از پرانتز استفاده کنید تا هم محاسبه صحیح باشد و هم کد خواناتر شود.

```
نقطه A: x=35.70, y=51.38

### تمرین ۲ — ارزیابی تصویر

یک تصویر ماهواره‌ای با مشخصات زیر داده شده است:

```

2. آیا NDVI نشان‌دهنده پوشش گیاهی متوسط است؟ (بین ۰.۳ تا ۰.۶)

عمق یک چاه حفاری `depth = 2500` متر است:
1. ۳۰۰ متر به عمق اضافه کنید
2. عمق را به کیلومتر تبدیل کنید (تقسیم بر ۱۰۰۰)
3. نتیجه را با دو رقم اعشار چاپ کنید

## ۳.۱۲ خلاصه

| دسته | عملگرها | کاربرد |
|---|---|---|
| حسابی | `+` `-` `*` `/` `//` `%` `**` | محاسبات عددی |
| مقایسه‌ای | `==` `!=` `>` `<` `>=` `<=` | مقایسه مقادیر |
| منطقی | `and` `or` `not` | ترکیب شرط‌ها |
| انتساب ترکیبی | `+=` `-=` `*=` `/=` `//=` `%=` `**=` | به‌روزرسانی متغیر |
| عضویت | `in` `not in` | بررسی وجود مقدار |

نکات کلیدی این فصل:
- عملگر `/` همیشه `float` برمی‌گرداند؛ برای نتیجه صحیح از `//` استفاده کنید
- `=` انتساب است، `==` مقایسه
- پرانتز بالاترین اولویت را دارد — در فرمول‌های پیچیده حتماً استفاده کنید
- عملگرهای منطقی (`and`, `or`, `not`) روی مقادیر بولین کار می‌کنند
