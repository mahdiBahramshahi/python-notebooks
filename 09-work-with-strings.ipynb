## ۹.۱ مقدمه

رشته (String) یکی از پرکاربردترین انواع داده در پایتون است. هر بار که با متن، نام، پیام یا داده‌های متنی کار می‌کنید، با رشته سروکار دارید.

در پایتون رشته دنباله‌ای **ترتیب‌دار** و **تغییرناپذیر** از کاراکترهاست. این یعنی می‌توانید به تک‌تک کاراکترها دسترسی داشته باشید، اما نمی‌توانید آن‌ها را مستقیم تغییر دهید.

در این فصل یاد می‌گیریم:
- تعریف رشته و دسترسی به کاراکترها
- متدهای پرکاربرد رشته
- قالب‌بندی رشته (f-string و format)
- جستجو و جایگزینی در رشته
- کار با رشته‌های چندخطی

## ۹.۴ ایندکس‌گذاری و برش‌زنی

رشته مثل لیست ایندکس دارد. ایندکس از صفر شروع می‌شود و ایندکس منفی از آخر می‌شمارد.
</div>


```
word = "Python"

print(word[0])    # P
print(word[3])    # h
print(word[-1])   # n
print(word[-3])   # h

# برش‌زنی
print(word[0:3])  # Pyt
print(word[2:])   # thon
print(word[:4])   # Pyth
print(word[::-1]) # nohtyP  (معکوس)
```

## ۹.۶ متدهای پرکاربرد رشته

### تغییر حروف
</div>


```
text = "Hello, World!"

print(text.upper())      # HELLO, WORLD!
print(text.lower())      # hello, world!
print(text.title())      # Hello, World!
print(text.swapcase())   # hELLO, wORLD!
print(text.capitalize()) # Hello, world!
```

<div dir="rtl" style="text-align:right;">

### حذف فاصله‌های اضافه
</div>


```
messy = "   Hello, World!   "

print(messy.strip())   # "Hello, World!"   (از هر دو طرف)
print(messy.lstrip())  # "Hello, World!   " (از چپ)
print(messy.rstrip())  # "   Hello, World!" (از راست)
```

<div dir="rtl" style="text-align:right;">

### جستجو در رشته
</div>


```
sentence = "The quick brown fox jumps over the lazy dog"

print(sentence.find("fox"))       # 16  (ایندکس اولین رخداد)
print(sentence.find("cat"))       # -1  (پیدا نشد)
print(sentence.count("the"))      # 1   (حساس به حروف بزرگ/کوچک)
print(sentence.count("the", 0, len(sentence)))  # 1
print(sentence.startswith("The")) # True
print(sentence.endswith("dog"))   # True
```

<div dir="rtl" style="text-align:right;">

## ۹.۷ جدول کامل متدهای پرکاربرد

| متد | کاربرد | مثال |
|---|---|---|
| `upper()` | همه حروف بزرگ | `"hi".upper()` → `"HI"` |
| `lower()` | همه حروف کوچک | `"HI".lower()` → `"hi"` |
| `title()` | اول هر کلمه بزرگ | `"hello world".title()` → `"Hello World"` |
| `strip()` | حذف فاصله دو طرف | `"  hi  ".strip()` → `"hi"` |
| `replace(a, b)` | جایگزینی `a` با `b` | `"cat".replace("c","b")` → `"bat"` |
| `split(sep)` | تقسیم به لیست | `"a,b".split(",")` → `["a","b"]` |
| `join(lst)` | ادغام لیست | `",".join(["a","b"])` → `"a,b"` |
| `find(s)` | ایندکس اولین رخداد | `"hello".find("l")` → `2` |
| `count(s)` | تعداد رخداد | `"hello".count("l")` → `2` |
| `startswith(s)` | آیا با `s` شروع می‌شود؟ | `"hi".startswith("h")` → `True` |
| `endswith(s)` | آیا با `s` تمام می‌شود؟ | `"hi".endswith("i")` → `True` |
| `isdigit()` | آیا همه کاراکترها عدد هستند؟ | `"123".isdigit()` → `True` |
| `isalpha()` | آیا همه کاراکترها حرف هستند؟ | `"abc".isalpha()` → `True` |
| `isspace()` | آیا همه کاراکترها فاصله هستند؟ | `"   ".isspace()` → `True` |
| `zfill(n)` | پر کردن با صفر از چپ | `"42".zfill(5)` → `"00042"` |
| `center(n)` | وسط‌چین در عرض `n` | `"hi".center(10)` → `"    hi    "` |

## ۹.۹ بررسی و اعتبارسنجی رشته
</div>


```
def validate_username(username):
    """بررسی می‌کند نام کاربری معتبر است یا نه."""
    if len(username) < 3:
        return False, "نام کاربری باید حداقل ۳ کاراکتر داشته باشد."
    if len(username) > 20:
        return False, "نام کاربری نباید بیشتر از ۲۰ کاراکتر داشته باشد."
    if not username[0].isalpha():
        return False, "نام کاربری باید با حرف شروع شود."
    if not username.replace("_", "").isalnum():
        return False, "نام کاربری فقط می‌تواند شامل حروف، اعداد و _ باشد."
    
    return True, "نام کاربری معتبر است."


test_names = ["Ali", "a", "123user", "valid_user_99", "this_name_is_way_too_long_for_sure"]

for uname in test_names:
    is_valid, msg = validate_username(uname)
    status = "✓" if is_valid else "✗"
    print(f"{status} '{uname}': {msg}")
```

خروجی:

✓ 'Ali': نام کاربری معتبر است.
✗ 'a': نام کاربری باید حداقل ۳ کاراکتر داشته باشد.
✗ '123user': نام کاربری باید با حرف شروع شود.
✓ 'valid_user_99': نام کاربری معتبر است.
✗ 'this_name_is_way_too_long_for_sure': نام کاربری نباید بیشتر از ۲۰ کاراکتر داشته باشد.

## ۹.۱۰ مثال کاربردی: پردازش داده متنی
</div>


```
def parse_csv_record(record):
    """یک رکورد CSV را پردازش و به دیکشنری تبدیل می‌کند."""
    parts = record.strip().split(",")
    if len(parts) != 4:
        return None

    return {
        "name":      parts[0].strip().title(),
        "latitude":  float(parts[1].strip()),
        "longitude": float(parts[2].strip()),
        "elevation": int(parts[3].strip())
    }


def format_station_info(station):
    """اطلاعات ایستگاه را به صورت خوانا فرمت می‌کند."""
    return (
        f"ایستگاه: {station['name']:<15} | "
        f"عرض: {station['latitude']:>8.3f} | "
        f"طول: {station['longitude']:>8.3f} | "
        f"ارتفاع: {station['elevation']:>5} m"
    )


# داده‌های خام
raw_data = [
    "  tehran  , 35.700, 51.380, 1191",
    "tabriz,38.080,46.290,1361",
    "  isfahan , 32.650 , 51.670 , 1590  ",
    "mashhad, 36.300, 59.600, 999",
]

print("گزارش ایستگاه‌های هواشناسی")
print("=" * 65)

stations = []
for record in raw_data:
    station = parse_csv_record(record)
    if station:
        stations.append(station)
        print(format_station_info(station))

    print("=" * 65)
    print(f"تعداد ایستگاه‌های پردازش‌شده: {len(stations)}")
```

ایستگاه: Tabriz          |  عرض:   38.080 |  طول:   46.290 | ارتفاع:  1361 m
ایستگاه: Isfahan         |  عرض:   32.650 |  طول:   51.670 | ارتفاع:  1590 m

## ۹.۱۱ نکات مهم

> **نکته ۱ — رشته تغییرناپذیر است:** متدهای رشته رشته اصلی را تغییر نمی‌دهند؛ یک رشته جدید برمی‌گردانند.
</div>


```
text = "hello"
text.upper()        # این کاری نمی‌کند!
print(text)         # hello — تغییر نکرد

text = text.upper() # باید نتیجه را ذخیره کنید
print(text)         # HELLO
```

<div dir="rtl" style="text-align:right;">

> **نکته ۲ — مقایسه رشته:** مقایسه حساس به حروف بزرگ/کوچک است. برای مقایسه بدون حساسیت، ابتدا `lower()` بگیرید.
</div>


```
a = "Python"
b = "python"

print(a == b)               # False
print(a.lower() == b.lower())  # True
```

<div dir="rtl" style="text-align:right;">

> **نکته ۳ — الحاق در حلقه:** الحاق رشته با `+` در حلقه کند است. از `join()` استفاده کنید.
</div>


```
words = ["Hello", "World", "Python"]

# روش کند (در حلقه‌های بزرگ)
result = ""
for w in words:
    result += w + " "

# روش بهتر
result = " ".join(words)
print(result)  # Hello World Python
```

<div dir="rtl" style="text-align:right;">

> **نکته ۴ — `find()` در مقابل `index()`:** هر دو ایندکس اولین رخداد را برمی‌گردانند، اما اگر رشته پیدا نشود، `find()` مقدار `-1` و `index()` خطا می‌دهد.
</div>


```
text = "Hello"
print(text.find("z"))   # -1
# print(text.index("z"))  # ValueError
```

## ۹.۱۲ تمرین‌های عملی

### تمرین ۱ — معکوس کردن کلمات
جمله‌ای از کاربر بگیرید. کلمات آن را معکوس کنید (نه کاراکترها) و چاپ کنید.
مثال: `"Hello World Python"` → `"Python World Hello"`

### تمرین ۲ — شمارش حروف
رشته‌ای از کاربر بگیرید. تعداد حروف بزرگ، کوچک، اعداد و فاصله‌ها را جداگانه بشمارید و چاپ کنید.

### تمرین ۳ — پالیندروم
تابعی بنویسید که بررسی کند یک کلمه پالیندروم است یا نه (از هر دو طرف یکسان خوانده می‌شود). مثال: `"racecar"`, `"level"`.

### تمرین ۴ — پردازش CSV ساده
رشته زیر را بگیرید و اطلاعات هر دانشجو را در یک دیکشنری ذخیره کنید، سپس نام دانشجویی با بالاترین نمره را چاپ کنید.

"Ali,85\nMaryam,92\nReza,74\nSara,98"

### تمرین ۵ — قالب‌بندی فاکتور
لیستی از تاپل‌های `(نام_کالا, قیمت, تعداد)` بسازید. با f-string یک فاکتور خوانا چاپ کنید که شامل ردیف‌های کالا و جمع کل باشد.
