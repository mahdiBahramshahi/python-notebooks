# فصل ۷: توابع در پایتون

## ۷.۱ مقدمه

کدنویسی تکراری خسته‌کننده و مستعد خطا است. **توابع** (Functions) به شما اجازه می‌دهند یک بلوک کد را یک بار بنویسید و هر بار که نیاز دارید آن را فراخوانی کنید. این کار کد را سازمان‌یافته‌تر، خواناتر و قابل استفاده مجدد می‌کند.

در این فصل یاد می‌گیریم:
- تابع چیست و چگونه تعریف می‌شود؟
- پارامترها و آرگومان‌ها
- مقدار بازگشتی (Return Value)
- انواع توابع (Built-in, User-defined)
- محدوده‌ی دید متغیرها (Scope)

---

## ۷.۲ تابع چیست؟

یک تابع مانند یک دستورالعمل یا یک "جعبه ابزار" کوچک است که وظیفه خاصی را انجام می‌دهد.
</div>


```
# ساختار کلی تعریف تابع
def function_name(parameters):
    # بدنه تابع
    return value  # اختیاری
```

<div dir="rtl" style="text-align:right;">

- `def` کلمه کلیدی برای تعریف تابع است.
- `function_name` باید مطابق قوانین نام‌گذاری متغیرها باشد (حروف کوچک، زیرخط به جای فاصله).
- `parameters` (اختیاری) ورودی‌هایی هستند که تابع دریافت می‌کند.
- `return` (اختیاری) مقداری را به بیرون تابع برمی‌گرداند.

### فراخوانی تابع (Calling a Function)
</div>


```
def greet():
    print("سلام!")

# فراخوانی تابع
greet()
greet()
```

## ۷.۴ مقدار بازگشتی (Return Value)

تابع می‌تواند نتیجه کار خود را با `return` به فراخواننده برگرداند.
</div>


```
def add_numbers(x, y):
    total = x + y
    return total


result = add_numbers(5, 3)
print(result)           # 8
print(add_numbers(10, -2))  # 8
```

## ۷.۵ Docstring — مستندسازی تابع

**Docstring** یک رشته توضیحی است که بلافاصله بعد از `def` می‌آید و هدف تابع را توضیح می‌دهد. نوشتن آن عادت خوبی است.
</div>


```
def calculate_ndvi(nir, red):
    """
    مقدار NDVI را بر اساس بازتابندگی باندهای NIR و Red محاسبه می‌کند.

    پارامترها:
    nir (float): بازتابندگی باند مادون قرمز نزدیک
    red (float): بازتابندگی باند قرمز

    خروجی:
    float یا None: مقدار NDVI یا None در صورت تقسیم بر صفر
    """
    if (nir + red) == 0:
        return None

    ndvi = (nir - red) / (nir + red)
    return ndvi


nir_val = 0.45
red_val = 0.12
ndvi_result = calculate_ndvi(nir_val, red_val)

if ndvi_result is not None:
    print(f"NDVI محاسبه شده: {ndvi_result:.4f}")
else:
    print("خطا: مخرج صفر بود.")
```

<div dir="rtl" style="text-align:right;">

خروجی:

NDVI محاسبه شده: 0.5789

فرمول NDVI:

NDVI = (NIR - Red)/(NIR + Red)

## ۷.۷ پارامترهای کلیدواژه‌ای (Keyword Arguments)

می‌توان آرگومان‌ها را با نام پارامتر مشخص کرد. این کار ترتیب ارسال را مهم نمی‌کند و خوانایی کد را بالا می‌برد.
</div>


```
def describe_location(name, latitude, longitude, elevation=0):
    print(f"نام:     {name}")
    print(f"عرض:     {latitude}")
    print(f"طول:     {longitude}")
    print(f"ارتفاع:  {elevation} متر")
```

```
describe_location("تهران", 35.70, 51.38, 1200)

print("-" * 25)

# فراخوانی با کلیدواژه (ترتیب مهم نیست)
describe_location(longitude=51.38, name="کرج", latitude=35.80, elevation=1300)

print("-" * 25)

# استفاده از مقدار پیش‌فرض برای ارتفاع
describe_location(name="شیراز", latitude=29.60, longitude=52.50)
```

## ۷.۸ بازگرداندن چند مقدار

یک تابع می‌تواند چند مقدار را با هم برگرداند. پایتون آن‌ها را به صورت یک **tuple** برمی‌گرداند.
</div>


```
import math


def calculate_stats(numbers):
    """محاسبه مجموع، میانگین و انحراف معیار یک لیست."""
    n = len(numbers)
    if n == 0:
        return None, None, None

    total = sum(numbers)
    mean = total / n
    variance = sum((x - mean) ** 2 for x in numbers) / n
    std_dev = math.sqrt(variance)

    return total, mean, std_dev


# داده‌های نمونه (بارش روزانه به میلی‌متر)
rainfall_mm = [5, 12, 0, 8, 15, 3, 0, 2, 10, 6]

total_rain, avg_rain, std_rain = calculate_stats(rainfall_mm)

if total_rain is not None:
    print(f"مجموع بارش:          {total_rain} میلی‌متر")
    print(f"میانگین بارش روزانه: {avg_rain:.2f} میلی‌متر")
    print(f"انحراف معیار بارش:   {std_rain:.2f} میلی‌متر")
else:
    print("داده‌ای برای محاسبه وجود ندارد.")
```

<div dir="rtl" style="text-align:right;">

خروجی:

مجموع بارش:          61 میلی‌متر

میانگین بارش روزانه: 6.10 میلی‌متر

انحراف معیار بارش:   4.78 میلی‌متر

فرمول انحراف معیار:

σ = sqrt(  ( Σ (xi - x̄)² )  /  n  )

## ۷.۹ محدوده‌ی دید متغیرها (Scope)

- **متغیر محلی (Local):** فقط درون تابع قابل دسترسی است.
- **متغیر سراسری (Global):** در کل برنامه قابل دسترسی است.
</div>


```
global_var = "من سراسری هستم"


def my_function():
    local_var = "من محلی هستم"
    print(local_var)
    print(global_var)  # متغیر سراسری از داخل تابع قابل خواندن است


my_function()
print(global_var)
# print(local_var)  # NameError: name 'local_var' is not defined
```

<div dir="rtl" style="text-align:right;">

### تغییر متغیر سراسری

برای تغییر (نه فقط خواندن) یک متغیر سراسری از داخل تابع، باید از `global` استفاده کنید:
</div>


```
count = 0


def increment_counter():
    global count
    count += 1
    print(f"شمارنده: {count}")


increment_counter()  # شمارنده: 1
increment_counter()  # شمارنده: 2
```

<div dir="rtl" style="text-align:right;">

> **توصیه:** تا حد امکان از تغییر متغیرهای سراسری خودداری کنید. بهتر است مقادیر را از طریق پارامتر دریافت و با `return` برگردانید. این کار کد را قابل پیش‌بینی‌تر و آسان‌تر برای اشکال‌زدایی می‌کند.

## ۷.۱۰ توابع داخلی (Built-in Functions)

پایتون توابع داخلی زیادی دارد که بدون نیاز به `import` قابل استفاده هستند:

| تابع | کاربرد | مثال |
|---|---|---|
| `print()` | نمایش خروجی | `print("سلام")` |
| `input()` | دریافت ورودی | `input("نام: ")` |
| `len()` | طول دنباله | `len([1, 2, 3])` → `3` |
| `type()` | نوع داده | `type(3.14)` → `float` |
| `int()`, `float()`, `str()` | تبدیل نوع | `int("5")` → `5` |
| `sum()`, `max()`, `min()` | محاسبات روی دنباله | `max([3, 1, 4])` → `4` |
| `abs()` | قدر مطلق | `abs(-7)` → `7` |
| `round()` | گرد کردن | `round(3.14159, 2)` → `3.14` |
| `range()` | تولید دنباله عددی | `range(1, 6)` |
| `sorted()` | مرتب‌سازی | `sorted([3, 1, 2])` → `[1, 2, 3]` |

## ۷.۱۱ مثال کاربردی: ارزیابی کیفیت تصویر ماهواره‌ای
</div>


```
def evaluate_image_quality(cloud_cover, ndvi, sun_elevation):
    """
    کیفیت تصویر ماهواره‌ای را بر اساس سه پارامتر ارزیابی می‌کند.

    پارامترها:
    cloud_cover (float): درصد پوشش ابر (0-100)
    ndvi (float): شاخص پوشش گیاهی (-1 تا 1)
    sun_elevation (float): زاویه ارتفاع خورشید (درجه)

    خروجی:
    str: سطح کیفیت تصویر
    """
    is_clear = cloud_cover < 10
    is_good_sun = sun_elevation > 30
    is_veg_present = ndvi > 0.2

    if is_clear and is_good_sun and is_veg_present:
        quality = "High"
    elif is_clear and is_good_sun:
        quality = "Medium"
    elif is_clear:
        quality = "Low (Sun Angle)"
    else:
        quality = "Poor (Cloud Cover)"

    return quality


# تست تابع با چند سناریو
scenarios = [
    {"cloud_cover": 5.0,  "ndvi": 0.7, "sun_elevation": 45.0},
    {"cloud_cover": 15.0, "ndvi": 0.6, "sun_elevation": 35.0},
    {"cloud_cover": 8.0,  "ndvi": 0.1, "sun_elevation": 20.0},
]

for i, scene in enumerate(scenarios, start=1):
    quality = evaluate_image_quality(**scene)
    print(f"سناریو {i}: {quality}")
```

<div dir="rtl" style="text-align:right;">

خروجی:

سناریو 1: High
سناریو 2: Poor (Cloud Cover)
سناریو 3: Low (Sun Angle)

## ۷.۱۲ تمرین‌های عملی

### تمرین ۱ — تابع محاسبه BMI
تابعی بنویسید که وزن (kg) و قد (m) را دریافت کند، BMI را محاسبه و برگرداند. سپس با مقادیر نمونه آن را فراخوانی کنید.

BMI = (weight)/(height^2)

### تمرین ۲ — تابع ارزیابی نمره
تابعی بنویسید که نمره دانشجو را بگیرد و وضعیت "قبول" یا "مردود" را برگرداند (نمره قبولی ۱۰).

### تمرین ۳ — تابع سلام
تابعی بنویسید که یک نام و یک پیام اختیاری (پیش‌فرض: "سلام") بگیرد و یک رشته خوش‌آمدگویی با f-string بسازد و برگرداند.

### تمرین ۴ — تابع بیشترین عدد
تابعی بنویسید که دو عدد بگیرد و بزرگ‌تر را برگرداند. (بدون استفاده از `max()`)

### تمرین ۵ — تابع آمار بارش
تابعی بنویسید که یک لیست از مقادیر بارش ماهانه بگیرد و مجموع، میانگین، بیشترین و کمترین مقدار را به صورت یک tuple برگرداند.

## ۷.۱۳ خلاصه

| مفهوم | نکته کلیدی |
|---|---|
| `def` | کلمه کلیدی تعریف تابع |
| پارامتر | متغیر در تعریف تابع |
| آرگومان | مقدار ارسالی هنگام فراخوانی |
| `return` | برگرداندن مقدار از تابع |
| `None` | مقدار بازگشتی پیش‌فرض (بدون return) |
| پارامتر پیش‌فرض | مقدار هنگام عدم ارسال آرگومان |
| Keyword Argument | ارسال آرگومان با نام پارامتر |
| Scope | محدوده دسترسی متغیر (Local / Global) |
| `global` | برای تغییر متغیر سراسری از داخل تابع |
| Docstring | مستندسازی هدف و رفتار تابع |

نکات کلیدی:
- هر تابع باید یک وظیفه مشخص داشته باشد (Single Responsibility)
- نام تابع باید گویا و فعلی باشد: `calculate_ndvi` نه `ndvi`
- پارامترهای پیش‌فرض همیشه بعد از پارامترهای اجباری می‌آیند
- به جای تغییر متغیر سراسری، از پارامتر و `return` استفاده کنید
- Docstring بنویسید — آینده‌ی شما ممنون خواهد بود!
