# نصب و راه‌اندازی محیط توسعه

## نصب پایتون

پایتون یک زبان برنامه‌نویسی متن‌باز و رایگان است. برای شروع کار، ابتدا باید آن را روی سیستم خود نصب کنید.

---

### نصب پایتون روی Windows

1. به سایت رسمی [python.org/downloads](https://www.python.org/downloads/) بروید
2. روی دکمه **"Download Python 3.x.x"** کلیک کنید
3. فایل `.exe` دانلود شده را اجرا کنید
4. در صفحه اول نصب، حتماً گزینه **"Add Python to PATH"** را تیک بزنید

> ⚠️ **مهم:** اگر این گزینه را تیک نزنید، پایتون در ترمینال شناخته نمی‌شود و باید مسیر را دستی تنظیم کنید.

5. روی **"Install Now"** کلیک کنید و منتظر بمانید
6. پس از نصب، ترمینال (CMD یا PowerShell) را باز کنید و تست کنید:
</div>


````
python --version
pip --version
````

<div dir="rtl" style="text-align:right;">

---

### نصب پایتون روی macOS

macOS معمولاً یک نسخه قدیمی پایتون ۲ دارد که **نباید از آن استفاده کنید**. پایتون ۳ را جداگانه نصب کنید:

**روش ۱ — دانلود مستقیم (ساده‌تر):**

1. به [python.org/downloads](https://www.python.org/downloads/) بروید
2. فایل `.pkg` مخصوص macOS را دانلود کنید
3. فایل را باز کرده و مراحل نصب را دنبال کنید
4. در ترمینال تست کنید:
</div>


````
python3 --version
pip3 --version
````

<div dir="rtl" style="text-align:right;">

**روش ۲ — از طریق Homebrew (پیشنهادی برای توسعه‌دهندگان):**

ابتدا Homebrew نصب کنید اگر ندارید:
</div>


````
/bin/bash -c "$(curl -fsSL https://raw.githubusercontent.com/Homebrew/install/HEAD/install.sh)"
````

<div dir="rtl" style="text-align:right;">

سپس پایتون نصب کنید:
</div>


````
brew install python
````

## محیط‌های برنامه‌نویسی (IDE و Text Editor)

برای نوشتن کد پایتون نیاز به یک محیط مناسب دارید. گزینه‌های مختلفی وجود دارد:

| محیط | نوع | مزایا | مناسب برای |
|---|---|---|---|
| **VS Code** | Text Editor | سبک، رایگان، قابل توسعه، Git داخلی | همه سطوح |
| **PyCharm** | IDE | قدرتمند، دیباگر حرفه‌ای، refactoring | پروژه‌های بزرگ |
| **Jupyter Notebook** | Notebook | تعاملی، نمایش نمودار در لحظه | علم داده، آموزش |
| **Spyder** | IDE | شبیه MATLAB، مناسب محاسبات علمی | محاسبات و تحقیق |
| **Sublime Text** | Text Editor | بسیار سبک و سریع | ویرایش سریع |
| **Google Colab** | آنلاین | بدون نصب، GPU رایگان | یادگیری و آزمایش |

> 💡 **پیشنهاد این دوره:** ما از **VS Code** استفاده می‌کنیم. سبک است، رایگان است، و با افزونه‌های مناسب به یک IDE کامل تبدیل می‌شود.

## نصب VS Code

### نصب VS Code روی Windows

1. به سایت [code.visualstudio.com](https://code.visualstudio.com) بروید
2. روی دکمه **"Download for Windows"** کلیک کنید
3. فایل نصبی `.exe` را دانلود و اجرا کنید
4. در مراحل نصب، گزینه‌های زیر را تیک بزنید:
   - `Add "Open with Code" action to Windows Explorer file context menu`
   - `Add "Open with Code" action to Windows Explorer directory context menu`
   - `Add to PATH`
5. نصب را تکمیل کنید و VS Code را باز کنید

---

### نصب VS Code روی macOS

1. به سایت [code.visualstudio.com](https://code.visualstudio.com) بروید
2. روی دکمه **"Download for macOS"** کلیک کنید
3. فایل `.zip` دانلود شده را باز کنید
4. فایل `Visual Studio Code.app` را به پوشه **Applications** بکشید
5. VS Code را از پوشه Applications اجرا کنید

برای دسترسی از ترمینال، کلید `Cmd+Shift+P` را بزنید و دستور زیر را اجرا کنید:

```
Shell Command: Install 'code' command in PATH
```
---

### نصب VS Code روی Linux

**Ubuntu / Debian:**
</div>


````
sudo apt update
sudo apt install software-properties-common apt-transport-https wget
wget -q https://packages.microsoft.com/keys/microsoft.asc -O- | sudo apt-key add -
sudo add-apt-repository "deb [arch=amd64] https://packages.microsoft.com/repos/vscode stable main"
sudo apt update
sudo apt install code
````

<div dir="rtl" style="text-align:right;">

**Fedora / RHEL:**
</div>


````
sudo rpm --import https://packages.microsoft.com/keys/microsoft.asc
sudo sh -c 'echo -e "[code]\nname=Visual Studio Code\nbaseurl=https://packages.microsoft.com/yumrepos/vscode\nenabled=1\ngpgcheck=1\ngpgkey=https://packages.microsoft.com/keys/microsoft.asc" > /etc/yum.repos.d/vscode.repo'
sudo dnf check-update
sudo dnf install code
````

<div dir="rtl" style="text-align:right;">

**Arch Linux:**
</div>


````
yay -S visual-studio-code-bin
````

## محیط مجازی (Virtual Environment) چیست و چرا نیاز داریم؟

### مشکل بدون venv

تصور کنید دو پروژه دارید:
- **پروژه A** به نسخه `numpy 1.21` نیاز دارد
- **پروژه B** به نسخه `numpy 1.26` نیاز دارد

اگر پکیج‌ها را مستقیم روی سیستم نصب کنید، این دو پروژه با هم **تداخل** پیدا می‌کنند و یکی از آن‌ها خراب می‌شود.

### راه‌حل: Virtual Environment

**venv** یک محیط پایتون ایزوله و مستقل برای هر پروژه می‌سازد. هر محیط:
- پکیج‌های مخصوص خودش را دارد
- نسخه مستقل از پکیج‌های دیگر دارد
- از سیستم اصلی و سایر پروژه‌ها جدا است
</div>


<div dir="rtl" style="text-align:right;">

> ✅ **قانون طلایی:** هر پروژه پایتون = یک venv جداگانه

## مدیریت پکیج‌ها با pip

**pip** ابزار رسمی نصب پکیج‌های پایتون است. وقتی داخل venv هستید، تمام پکیج‌هایی که نصب می‌کنید فقط در همان محیط ذخیره می‌شوند.

### دستورات اصلی pip

نصب یک پکیج:
</div>


````
pip install numpy
````

<div dir="rtl" style="text-align:right;">

نصب نسخه خاص:
</div>


````
pip install numpy==1.26.0
````

<div dir="rtl" style="text-align:right;">

حذف یک پکیج:
</div>


````
pip uninstall numpy
````

<div dir="rtl" style="text-align:right;">

نمایش لیست پکیج‌های نصب شده:
</div>


````
pip list
````

<div dir="rtl" style="text-align:right;">

ذخیره لیست پکیج‌ها در فایل (برای اشتراک‌گذاری پروژه):
</div>


````
pip freeze > requirements.txt
````

<div dir="rtl" style="text-align:right;">

نصب همه پکیج‌ها از فایل requirements.txt:
</div>


````
pip install -r requirements.txt
````

<div dir="rtl" style="text-align:right;">

آپدیت یک پکیج:
</div>


````
pip install --upgrade numpy
````

<div dir="rtl" style="text-align:right;">

جستجوی اطلاعات یک پکیج:
</div>


````
pip show numpy
````

<div dir="rtl" style="text-align:right;">

### مثال عملی: نصب پکیج‌های پرکاربرد

مطمئن شوید داخل venv هستید: `(venv)` باید اول خط باشد
</div>


````
pip install numpy pandas matplotlib
````

<div dir="rtl" style="text-align:right;">

> 📦 **pip چیست؟** pip مخفف **P**ip **I**nstalls **P**ackages است. پکیج‌ها را از مخزن رسمی [PyPI](https://pypi.org) دانلود و نصب می‌کند. PyPI در حال حاضر بیش از ۵۰۰ هزار پکیج دارد.

### مرحله ۲: انتخاب Interpreter پایتون (venv)

این مرحله **مهم‌ترین** قدم است. VS Code باید بداند از کدام پایتون (کدام venv) استفاده کند.

1. کلید میانبر `Ctrl+Shift+P` (یا `Cmd+Shift+P` در macOS) را بزنید
2. عبارت `Python: Select Interpreter` را تایپ کنید
3. از لیست، گزینه‌ای که مسیر `venv` پروژه شما را دارد انتخاب کنید

مسیر معمولاً اینگونه است:
- **Windows:** `.\venv\Scripts\python.exe`
- **macOS/Linux:** `./venv/bin/python`

> 💡 اگر venv در لیست نبود، روی **"Enter interpreter path"** کلیک کنید و مسیر را دستی وارد کنید.

---

### مرحله ۳: تأیید راه‌اندازی

پس از انتخاب interpreter، در پایین صفحه VS Code نام محیط انتخاب شده نمایش داده می‌شود:
</div>


<div dir="rtl" style="text-align:right;">

`Python 3.x.x ('venv': venv)`

یک فایل جدید بسازید به نام `main.py` و این کد را بنویسید:
</div>


````
%%python
import sys

print("پایتون راه‌اندازی شد!")
print(f"نسخه پایتون: {sys.version}")
print(f"مسیر اجرا: {sys.executable}")
````

<div dir="rtl" style="text-align:right;">

با زدن کلید `F5` یا کلیک روی دکمه ▶ در بالا-راست، کد را اجرا کنید.

- یا از منو: `Terminal > New Terminal`

وقتی ترمینال باز می‌شود، باید `(venv)` را در ابتدای خط ببینید — یعنی همه چیز درست است.
