## تابع `print()` را بیشتر بشناسیم

`print()` یکی از پرکاربردترین توابع پایتون است. کار این تابع ساده است: **هر چیزی که داخل پرانتزش بنویسید، روی صفحه چاپ می‌کند.**

متن، عدد، نتیجه یک محاسبه — هر چیزی باشد، `print()` آن را نشان می‌دهد.
</div>


```python
print("این یک متن است")
print(42)
print(3.14)
print(10 + 5)
```

### تغییر پایان‌دهنده با `end`

به‌طور پیش‌فرض `print()` بعد از هر چاپ به خط جدید می‌رود. با پارامتر `end` می‌توانید این رفتار را عوض کنید:
</div>


```python
print("زمین‌شناسی", end=" | ")
print("سنجش از دور", end=" | ")
print("پایتون")
```

### یک مثال کامل‌تر

بیایید همه چیزهایی که یاد گرفتیم را در یک برنامه ترکیب کنیم:
</div>


```python
print("=" * 40)
print("  اطلاعات ایستگاه لرزه‌نگاری")
print("=" * 40)

print("نام ایستگاه: تهران")
print("کد ایستگاه: THR")
print("عرض جغرافیایی: 35.7018 درجه شمالی")
print("طول جغرافیایی: 51.3890 درجه شرقی")
print("ارتفاع: 1191 متر از سطح دریا")

print("-" * 40)

print("آخرین زمین‌لرزه ثبت‌شده:")
print("  بزرگا: 4.2 ریشتر")
print("  عمق کانون: 15 کیلومتر")
print("  منطقه: البرز مرکزی")

print("=" * 40)
print("پایان گزارش")
```

## کامنت‌گذاری (Comments)

**کامنت‌ها** خطوطی هستند که پایتون آن‌ها را **اجرا نمی‌کند**. از کامنت‌ها برای توضیح کد استفاده می‌کنیم — هم برای خودمان در آینده، هم برای دیگرانی که کد را می‌خوانند.

### کامنت تک‌خطی

با علامت `#` شروع می‌شود:
</div>


<div dir="rtl" style="text-align:right;">

# این برنامه اطلاعات یک نمونه سنگ را نمایش می‌دهد

print("نوع سنگ: گرانیت")    # کامنت می‌تواند انتهای خط هم باشد
print("سن: 65 میلیون سال")

### کامنت چندخطی

با سه علامت نقل‌قول:
</div>


```python
"""
این برنامه برای نمایش مختصات
یک منطقه مطالعاتی نوشته شده است.
منطقه: البرز مرکزی
"""

print("عرض جغرافیایی: 36.0 درجه شمالی")
print("طول جغرافیایی: 51.5 درجه شرقی")
```

<div dir="rtl" style="text-align:right;">

> ✅ **کامنت‌گذاری خوب** نشانه یک برنامه‌نویس حرفه‌ای است. همیشه کدتان را توضیح دهید.

## خطاها در پایتون

اشتباه کردن بخشی **طبیعی** از یادگیری برنامه‌نویسی است! وقتی کد اشتباهی بنویسید، پایتون یک **پیام خطا** نشان می‌دهد.

### انواع رایج خطا

| نوع خطا | توضیح | مثال |
|---|---|---|
| **SyntaxError** | اشتباه در نوشتن دستور | فراموش کردن پرانتز |
| **NameError** | استفاده از متغیر تعریف‌نشده | اشتباه تایپی در نام |
| **TypeError** | عملیات اشتباه روی نوع داده | جمع عدد با متن |
| **IndentationError** | مشکل در فاصله‌گذاری | تورفتگی اشتباه |

### مثال: خطای رایج
</div>


```python
# کد اشتباه — پرانتز بسته نشده
print("سلام"
```

<div dir="rtl" style="text-align:right;">

SyntaxError: '(' was never closed

# کد درست
</div>


```python
print("سلام")
```

<div dir="rtl" style="text-align:right;">

### چطور خطاها را رفع کنیم؟

1. **پیام خطا را بخوانید** — پایتون دقیقاً می‌گوید مشکل کجاست
2. **شماره خط را ببینید** — خطا در کدام خط رخ داده؟
3. **نوع خطا را شناسایی کنید** — SyntaxError؟ NameError؟
4. **جستجو کنید** — Stack Overflow و مستندات پایتون دوست شما هستند

> 🚀 از خطاها نترسید! هر برنامه‌نویس حرفه‌ای روزانه با خطاهای زیادی مواجه می‌شود. خطا یعنی پیشرفت!

### تمرین ۲ — متوسط

- تاریخ تصویربرداری
- قدرت تفکیک مکانی (متر)
- باندهای طیفی موجود

با استفاده از کاراکترهای `*`، `-` و `|` یک جدول ساده برای نمایش مشخصات ۳ نوع سنگ (آذرین، رسوبی، دگرگونی) طراحی و چاپ کنید — فقط با `print()` و بدون هیچ ابزار دیگری.
