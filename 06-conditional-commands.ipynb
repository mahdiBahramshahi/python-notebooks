## ۵.۱ مقدمه

در دنیای واقعی، تصمیم‌گیری بخش جدایی‌ناپذیر هر برنامه‌ای است. آیا کاربر وارد شده؟ آیا عدد وارد‌شده معتبر است؟ آیا دانشجو قبول شده؟

پایتون با ساختار `if` به برنامه اجازه می‌دهد بر اساس شرایط مختلف، مسیرهای متفاوتی را طی کند.

در این فصل یاد می‌گیریم:
- ساختار `if`, `elif`, `else`
- شرط‌های تودرتو (Nested Conditions)
- عملگر سه‌تایی (Ternary Operator)
- ترکیب شرط‌ها با `and`, `or`, `not`
- کاربردهای عملی تصمیم‌گیری

---

## ۵.۲ ساختار if ساده

ساده‌ترین شکل تصمیم‌گیری: اگر شرطی برقرار بود، یک کار انجام بده.
</div>


```
if شرط:
    # کدی که اجرا می‌شود اگر شرط True باشد
```

```
temperature = 38.5

if temperature > 37.5:
    print("تب دارید، استراحت کنید.")
```

<div dir="rtl" style="text-align:right;">

خروجی:

تب دارید، استراحت کنید.

> **نکته مهم:** پایتون از **indentation** (تورفتگی) برای مشخص کردن بلوک کد استفاده می‌کند. معمولاً ۴ فاصله یا یک Tab استفاده می‌شود.

---

## ۵.۳ ساختار if-else

اگر شرط برقرار بود یک کار، در غیر این صورت کار دیگری انجام بده.
</div>


```
if شرط:
    # اجرا می‌شود اگر شرط True باشد
else:
    # اجرا می‌شود اگر شرط False باشد
```

```
score = 14

if score >= 10:
    print("قبول شدید!")
else:
    print("متأسفانه مردود شدید.")
```

## ۵.۴ ساختار if-elif-else

برای بررسی چند شرط مختلف به ترتیب:
</div>


```
if شرط_اول:
    # اجرا می‌شود اگر شرط اول True باشد
elif شرط_دوم:
    # اجرا می‌شود اگر شرط دوم True باشد
elif شرط_سوم:
    # اجرا می‌شود اگر شرط سوم True باشد
else:
    # اجرا می‌شود اگر هیچ‌کدام True نباشد
```

```
score = int(input("نمره خود را وارد کنید: "))

if score >= 18:
    grade = "عالی"
elif score >= 15:
    grade = "خوب"
elif score >= 12:
    grade = "متوسط"
elif score >= 10:
    grade = "قبولی"
else:
    grade = "مردود"

print(f"وضعیت شما: {grade}")
```

اجرا:

وضعیت شما: خوب

> **نکته:** پایتون شرط‌ها را از بالا به پایین بررسی می‌کند و به محض اینکه یکی True شد، بقیه را نادیده می‌گیرد.

## ۵.۶ شرط‌های تودرتو (Nested Conditions)

می‌توان یک `if` را داخل `if` دیگری قرار داد:
</div>


```
username = input("نام کاربری: ")
password = input("رمز عبور: ")

if username == "admin":
    if password == "1234":
        print("خوش آمدید!")
    else:
        print("رمز عبور اشتباه است.")
else:
    print("نام کاربری یافت نشد.")
```

## ۵.۷ عملگر سه‌تایی (Ternary Operator)

برای شرط‌های ساده می‌توان همه چیز را در یک خط نوشت:
</div>


```
مقدار_اگر_True if شرط else مقدار_اگر_False
```

```
score = 15
result = "قبول" if score >= 10 else "مردود"
print(result)    # قبول
```

```
temperature = 25
weather = "گرم" if temperature > 30 else "معتدل"
print(f"هوا {weather} است.")    # هوا معتدل است.
```

<div dir="rtl" style="text-align:right;">

این روش برای موارد ساده خوانا است، اما برای شرط‌های پیچیده از `if-elif-else` معمولی استفاده کنید.

| `None` | NoneType | False |
| هر مقدار دیگر | — | True |

## ۵.۹ مثال کاربردی: سیستم ثبت‌نام دوره
</div>


```
print("=== سیستم ثبت‌نام دوره آموزشی ===\n")

name = input("نام و نام خانوادگی: ")
age = int(input("سن: "))
has_prerequisite = input("آیا پیش‌نیاز را گذرانده‌اید؟ (بله/خیر): ")
gpa = float(input("معدل (از ۲۰): "))

is_eligible_age = age >= 18
has_prereq = has_prerequisite == "بله"
is_eligible_gpa = gpa >= 12.0

print(f"\n--- نتیجه بررسی برای {name} ---")

if is_eligible_age and has_prereq and is_eligible_gpa:
    if gpa >= 17:
        status = "پذیرفته شده با بورسیه"
    elif gpa >= 14:
        status = "پذیرفته شده"
    else:
        status = "پذیرفته شده با شرط"
else:
    if not is_eligible_age:
        status = "رد شده — سن کمتر از ۱۸ سال"
    elif not has_prereq:
        status = "رد شده — پیش‌نیاز گذرانده نشده"
    else:
        status = "رد شده — معدل کافی نیست"

print(f"وضعیت: {status}")
```

## ۵.۱۰ مثال کاربردی: ماشین حساب پیشرفته
</div>


```
print("=== ماشین حساب ===")
print("عملگرهای مجاز: +  -  *  /  //  %  **\n")

num1 = float(input("عدد اول: "))
operator = input("عملگر: ")
num2 = float(input("عدد دوم: "))

result = None

if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        print("خطا: تقسیم بر صفر!")
elif operator == "//":
    if num2 != 0:
        result = num1 // num2
    else:
        print("خطا: تقسیم بر صفر!")
elif operator == "%":
    if num2 != 0:
        result = num1 % num2
    else:
        print("خطا: تقسیم بر صفر!")
elif operator == "**":
    result = num1 ** num2
else:
    print(f"خطا: عملگر '{operator}' شناخته نشد.")

if result is not None:
    print(f"\nنتیجه: {num1} {operator} {num2} = {result:,.4f}")
```

<div dir="rtl" style="text-align:right;">

اجرا:

=== ماشین حساب ===

عملگرهای مجاز: +  -  *  /  //  %  **

عدد اول: 144

عملگر: **

عدد دوم: 0.5

نتیجه: 144.0 ** 0.5 = 12.0000

## ۵.۱۲ تمرین‌های عملی

### تمرین ۱ — طبقه‌بندی BMI
برنامه‌ای بنویسید که وزن (kg) و قد (m) را از کاربر بگیرد، BMI را محاسبه کند و وضعیت را نمایش دهد:

| BMI | وضعیت |
|---|---|
| کمتر از ۱۸.۵ | کمبود وزن |
| ۱۸.۵ تا ۲۴.۹ | وزن طبیعی |
| ۲۵ تا ۲۹.۹ | اضافه وزن |
| ۳۰ و بیشتر | چاقی |

فرمول: BMI = weight / (height ** 2)

### تمرین ۲ — تشخیص سال کبیسه
برنامه‌ای بنویسید که یک سال میلادی را از کاربر بگیرد و بگوید کبیسه است یا نه.

قانون: سال کبیسه است اگر بر ۴ بخش‌پذیر باشد، به جز سال‌هایی که بر ۱۰۰ بخش‌پذیرند — مگر اینکه بر ۴۰۰ هم بخش‌پذیر باشند.

### تمرین ۳ — بازی حدس عدد
برنامه‌ای بنویسید که یک عدد ثابت (مثلاً ۴۲) در ذهن داشته باشد، کاربر حدس بزند، و برنامه بگوید بزرگ‌تر، کوچک‌تر یا درست است.

## ۵.۱۳ خلاصه

| ساختار | کاربرد |
|---|---|
| `if` | بررسی یک شرط |
| `if-else` | دو مسیر: True یا False |
| `if-elif-else` | چند مسیر مختلف |
| `if` تودرتو | شرط داخل شرط |
| `x if cond else y` | شرط یک‌خطی ساده |

نکات کلیدی:
- indentation در پایتون اجباری و معنادار است
- شرط‌ها از بالا به پایین بررسی می‌شوند، اولین True اجرا می‌شود
- رشته خالی، صفر و None در شرط‌ها False محسوب می‌شوند
- برای مقایسه چند مقدار از `in` استفاده کنید
- `==` برای مقایسه، `=` برای انتساب
