## ۲.۱ مقدمه

در فصل قبل یاد گرفتیم چطور پایتون را نصب کنیم و اولین برنامه را بنویسیم. حالا وقت آن است که با **متغیرها** و **انواع داده** آشنا شویم — این دو مفهوم، ستون فقرات هر برنامه‌ای هستند.

در این فصل یاد می‌گیریم:
- متغیر چیست و چطور تعریف می‌شود؟

- چطور نوع داده را بررسی و تبدیل کنیم؟
- قوانین نام‌گذاری متغیرها

## ۲.۲ متغیر چیست؟

یک **متغیر** (Variable) مثل یک جعبه است که می‌توانید یک مقدار را داخل آن بگذارید و بعداً از آن استفاده کنید.

در پایتون برای تعریف متغیر **نیازی به اعلان نوع** ندارید — فقط کافی است اسم بگذارید و مقدار بدهید:
</div>

## ۲.۳ انواع داده در پایتون

پایتون چند نوع داده پایه دارد:

| نوع داده | نام انگلیسی | مثال |
|---|---|---|
| عدد صحیح | int | 42 |
| عدد اعشاری | float | 3.14 |
| رشته متنی | str | "Basalt" |
| بولین | bool | True / False |
| خالی | NoneType | None |

## ۲.۴ عدد صحیح (int)

اعداد صحیح شامل اعداد مثبت، منفی و صفر هستند — **بدون اعشار**.
</div>


```python
age = 25
depth = 2500
count = -10
year = 2024

print(age)
print(depth)
print(count)
print(year)
```

خروجی:

-10

برای خوانایی بهتر اعداد بزرگ، می‌توانید از `_` به عنوان جداکننده استفاده کنید:
</div>


```python
earth_mass = 5_972_000_000_000_000_000_000_000
distance = 149_597_870_700

print(earth_mass)
print(distance)
```

### f-string — بهترین روش ترکیب رشته‌ها
</div>


```python
satellite = "Landsat-8"
resolution = 30
date = "2024-03-15"

info = f"ماهواره: {satellite} | رزولوشن: {resolution} | تاریخ: {date}"
print(info)
```

## ۲.۱۳ مثال کاربردی: پروفایل یک تصویر ماهواره‌ای
</div>


<div dir="rtl" style="text-align:right;">

# اطلاعات تصویر
</div>


```python
scene_id = "LC08_L2SP_164034_20230815"
satellite = "Landsat-8"
date = "2023-08-15"
cloud_cover = 3.2
pixel_size = 30.0
```

<div dir="rtl" style="text-align:right;">

# مقادیر بازتابندگی
</div>


```python
band_red = 0.103
band_nir = 0.312
```

<div dir="rtl" style="text-align:right;">

# محاسبه NDVI
</div>


```python
ndvi = (band_nir - band_red) / (band_nir + band_red)
```

<div dir="rtl" style="text-align:right;">

# گزارش
</div>


```python
print("=" * 45)
print(f"  ماهواره : {satellite}")
print(f"  تاریخ   : {date}")
print(f"  شناسه   : {scene_id}")
print(f"  پوشش ابر: {cloud_cover}%")
print(f"  رزولوشن : {pixel_size}m")
print("-" * 45)
print(f"  Red  : {band_red}")
print(f"  NIR  : {band_nir}")
print(f"  NDVI : {ndvi:.4f}")
print("=" * 45)
```

## ۲.۱۴ تمرین‌های عملی

### تمرین ۱
متغیرهای زیر را تعریف کرده و نوع هر کدام را با `type()` چاپ کنید:
- نام یک کوه
- ارتفاع آن به متر
- آیا قله برفی دارد؟
- تاریخ آخرین صعود (رشته)

### تمرین ۲
دو متغیر `band_green = 0.118` و `band_nir = 0.287` تعریف کنید، شاخص NDWI را محاسبه و با f-string چاپ کنید:
</div>


<div dir="rtl" style="text-align:right;">

NDWI = (Green - NIR) / (Green + NIR)

### تمرین ۳
اطلاعات یک چاه حفاری (نام، عمق، دما، فشار، وضعیت تولید) را در متغیرهای مناسب ذخیره کرده و یک گزارش کامل با f-string چاپ کنید.
