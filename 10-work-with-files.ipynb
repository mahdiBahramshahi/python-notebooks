## ۱۰.۱ مقدمه

در فصل‌های قبل، داده‌ها را در متغیرها ذخیره یا از کاربر دریافت می‌کردیم. اما این داده‌ها با پایان برنامه از بین می‌روند. برای ذخیره دائمی اطلاعات، نیاز به کار با **فایل‌ها** داریم.

فایل‌ها می‌توانند حاوی متن (مثل فایل‌های `.txt` یا `.csv`) یا داده‌های باینری (مثل عکس‌ها و ویدئوها) باشند. پایتون ابزارهای قدرتمندی برای خواندن و نوشتن در فایل‌ها فراهم می‌کند.

در این فصل یاد می‌گیریم:
- باز کردن و بستن فایل‌ها
- خواندن محتوای فایل
- نوشتن در فایل‌ها
- کار با حالت‌های مختلف فایل
- استفاده از `with open()` برای مدیریت خودکار

---

## ۱۰.۲ باز کردن و بستن فایل

برای کار با یک فایل، ابتدا باید آن را **باز** کنید. این کار با تابع `open()` انجام می‌شود.

# ساختار کلی
</div>


````python
file_object = open(filename, mode)
````

<div dir="rtl" style="text-align:right;">

- `filename`: نام فایل (مثلاً `"data.txt"`)
- `mode`: حالتی که فایل را باز می‌کنید (مثلاً `'r'` برای خواندن)

پس از اتمام کار، **حتماً** باید فایل را **ببندید** تا منابع آزاد شوند و داده‌ها ذخیره شوند.
</div>


````python
file = open("my_notes.txt", "w")  # باز کردن برای نوشتن
    # ... اینجا کارهایی روی فایل انجام می‌دهیم ...
file.close()  # بستن فایل
````

<div dir="rtl" style="text-align:right;">

## ۱۰.۳ حالت‌های باز کردن فایل (Modes)

| حالت | شرح |
|---|---|
| `'r'` | **Read (خواندن)**: پیش‌فرض. فایل را برای خواندن باز می‌کند. خطا می‌دهد اگر فایل وجود نداشته باشد. |
| `'w'` | **Write (نوشتن)**: فایل را برای نوشتن باز می‌کند. اگر فایل وجود داشته باشد، محتوای قبلی پاک می‌شود. اگر وجود نداشته باشد، ساخته می‌شود. |
| `'a'` | **Append (اضافه کردن)**: محتوا به انتهای فایل اضافه می‌شود. اگر فایل وجود نداشته باشد، ساخته می‌شود. |
| `'x'` | **Create (ساختن)**: فایل را برای ساختن باز می‌کند. اگر فایل وجود داشته باشد، خطا می‌دهد. |
| `'b'` | **Binary (باینری)**: برای کار با فایل‌های غیرمتنی. با حالت‌های دیگر ترکیب می‌شود (مثلاً `'rb'` یا `'wb'`). |
| `'+'` | **Update (خواندن و نوشتن)**: با حالت‌های دیگر ترکیب می‌شود (مثلاً `'r+'` یا `'w+'`). |

### ۲. خواندن خط به خط با `readline()`
</div>


````python
file = open("data/sample.txt", "r")
line1 = file.readline()
line2 = file.readline()
print(f"خط اول: {line1.strip()}")  # strip() برای حذف \n آخر خط
print(f"خط دوم: {line2.strip()}")
file.close()
````

### ۳. خواندن همه خطوط در لیست با `readlines()`

````python
file = open("data/sample.txt", "r")
lines = file.readlines()
print(lines)
file.close()
````

### پیمایش فایل با حلقه `for` (روش پیشنهادی)

این روش حافظه کمتری مصرف می‌کند چون کل فایل را یکجا بارگذاری نمی‌کند.
</div>


````python
file = open("data/sample.txt", "r")
print("محتوای فایل:")
for line in file:
    print(f"  -> {line.strip()}")
    file.close()
````

<div dir="rtl" style="text-align:right;">

خروجی:

  -> Line 1: Hello World
  -> Line 2: This is a sample file.
  -> Line 3: For reading demonstration.

## ۱۰.۵ نوشتن در فایل

### حالت `'w'` (Write) — بازنویسی
</div>


<div dir="rtl" style="text-align:right;">

# اگر فایل موجود باشد، پاک شده و از نو نوشته می‌شود
</div>


````python
with open("output.txt", "w") as f:
    f.write("This is the first line.\n")
    f.write("This overwrites the previous content.\n")
````

<div dir="rtl" style="text-align:right;">

### حالت `'a'` (Append) — اضافه کردن
</div>


<div dir="rtl" style="text-align:right;">

# محتوا به انتهای فایل اضافه می‌شود
</div>


````python
with open("output.txt", "a") as f:
    f.write("This line is appended.\n")
````

<div dir="rtl" style="text-align:right;">

### نوشتن چند خط با `writelines()`

`writelines()` یک لیست از رشته‌ها را می‌گیرد و در فایل می‌نویسد.

> **توجه:** `writelines()` خودش `\n` اضافه نمی‌کند. باید در هر رشته آن را بگنجانید.
</div>


````python
lines_to_write = ["First line\n", "Second line\n", "Third line\n"]
with open("output_lines.txt", "w") as f:
    f.writelines(lines_to_write)
````

<div dir="rtl" style="text-align:right;">

## ۱۰.۶ استفاده از `with open()` — بهترین روش

`with open(...) as ...:` تضمین می‌کند که فایل **حتماً** پس از اتمام بلوک کد بسته شود، حتی اگر خطا رخ دهد.
</div>


````python
try:
    with open("sample.txt", "r") as infile:
        content = infile.read()
        print("فایل خوانده شد.")

    with open("output.txt", "a") as outfile:
        outfile.write("Appended using 'with'.\n")
        print("به فایل اضافه شد.")

except FileNotFoundError:
    print("خطا: فایل sample.txt یافت نشد.")
except Exception as e:
    print(f"خطایی رخ داد: {e}")
````

## ۱۰.۷ کار با فایل‌های متنی و باینری

- **فایل‌های متنی (`'r'`, `'w'`, `'a'`):** پایتون به طور خودکار کاراکترهای پایان خط (`\n`) را مدیریت می‌کند.
- **فایل‌های باینری (`'rb'`, `'wb'`, `'ab'`):** برای داده‌هایی که متن نیستند (عکس، فایل فشرده، داده‌های خام). `read()` بایت برمی‌گرداند و `write()` بایت می‌گیرد.
</div>

````python
try:
    with open("data/image.jpeg", "rb") as img_file:
        image_data = img_file.read()
        print(f"اندازه تصویر: {len(image_data)} بایت")
except FileNotFoundError:
    print("فایل image.jpg پیدا نشد.")
````

## ۱۰.۸ تعیین انکدینگ (Encoding)

## ۱۰.۹ مثال کاربردی: خواندن و پردازش فایل CSV

فرض کنید فایلی به نام `data.csv` داریم:
</div>


<div dir="rtl" style="text-align:right;">

> همانند فایل txt که ساختیم، یک فایل به نام `data.csv` بسازید و محتوای زیر را در آن قرار دهید.
</div>


<div dir="rtl" style="text-align:right;">

```csv
City,Latitude,Longitude,Elevation
Tehran,35.70,51.38,1191
Tabriz,38.08,46.29,1361
Isfahan,32.65,51.67,1590
```
</div>


````python
import csv  # ماژول مخصوص CSV

stations_data = []

try:
    with open("data/data1.csv", "r", encoding="utf-8") as csvfile:
        # DictReader هر ردیف را به صورت دیکشنری می‌خواند
        reader = csv.DictReader(csvfile)

        for row in reader:
            row['Latitude']  = float(row['Latitude'])
            row['Longitude'] = float(row['Longitude'])
            row['Elevation'] = int(row['Elevation'])
            stations_data.append(row)

    print("داده‌های CSV خوانده شد:")
    for station in stations_data:
        print(station)

    # یافتن بلندترین ایستگاه
    highest = max(stations_data, key=lambda s: s["Elevation"])
    print(f"\nبلندترین ایستگاه: {highest['City']} ({highest['Elevation']} m)")

except FileNotFoundError:
    print("خطا: فایل data.csv یافت نشد.")
except Exception as e:
    print(f"خطا در پردازش CSV: {e}")
````

خروجی:

داده‌های CSV خوانده شد:

{'City': 'Tabriz', 'Latitude': 38.08, 'Longitude': 46.29, 'Elevation': 1361}
{'City': 'Isfahan', 'Latitude': 32.65, 'Longitude': 51.67, 'Elevation': 1590}

بلندترین ایستگاه: Isfahan (1590 m)

## ۱۰.۱۰ مثال کاربردی: کپی کردن فایل
</div>


````python
source_file      = "sample.txt"
destination_file = "sample_copy.txt"

try:
    with open(source_file, "rb") as infile, open(destination_file, "wb") as outfile:
        chunk_size = 4096  # 4KB

        while True:
            chunk = infile.read(chunk_size)
            if not chunk:
                break  # پایان فایل
            outfile.write(chunk)

    print(f"فایل '{source_file}' با موفقیت به '{destination_file}' کپی شد.")

except FileNotFoundError:
    print(f"خطا: فایل منبع '{source_file}' یافت نشد.")
except Exception as e:
    print(f"خطای کپی فایل: {e}")
````

## ۱۰.۱۱ نکات مهم

> **نکته ۱ — همیشه از `with open()` استفاده کنید:** این روش تضمین می‌کند فایل حتی در صورت بروز خطا بسته شود.

> **نکته ۲ — حالت `'w'` خطرناک است:** اگر فایل از قبل وجود داشته باشد، تمام محتوای آن بدون هشدار پاک می‌شود. قبل از استفاده مطمئن شوید.

> **نکته ۳ — انکدینگ را فراموش نکنید:** برای فایل‌های حاوی متن فارسی یا کاراکترهای غیر ASCII، همیشه `encoding="utf-8"` را مشخص کنید.

> **نکته ۴ — برای فایل‌های بزرگ از حلقه استفاده کنید:** به جای `read()` که کل فایل را در حافظه بارگذاری می‌کند، با حلقه `for` خط به خط بخوانید.
