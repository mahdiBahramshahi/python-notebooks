## چرا پایتون؟
پایتون از محبوب‌ترین زبان‌های برنامه‌نویسی جهانه، و دلیلش کاملاً ساده‌ست:
- یادگیریش آسونه
- کدهاش خواناست
- ابزار و کتابخانه برای هر کاری داره
- جامعه کاربری خیلی بزرگی پشتشه

اگر اولین زبان برنامه‌نویسی‌ات هست، عالیه—چون پایتون دقیقاً همون زبانیه که خیلی از دانشجوها، محقق‌ها و مهندس‌ها برای **شروع** انتخاب می‌کنن.

## پایتون و علوم زمین؟
این دوره تخصصی علوم زمین نیست، اما چون مخاطبان سایت زمین‌سنج معمولاً در رشته‌هایی مثل زمین‌شناسی، ژئوفیزیک، GIS، سنجش‌ازدور و آب‌زمین‌شناسی فعالیت دارن، خیلی مختصر اشاره کنیم که پایتون در این حوزه‌ها چه کاربردهایی داره.

### کاربردهای پایتون در علوم‌زمین (در حد آشنایی)
- پردازش تصاویر ماهواره‌ای
- تحلیل داده‌های چاه و لاگ
- نقشه‌سازی با GIS
- تحلیل لرزه‌ای و پردازش سیگنال
- ساخت مدل‌های زمین‌شناسی یا آب زیرزمینی
- تحلیل داده‌های سری‌زمانی مثل بارش، دبی و تراز آب

ما در این دوره صرفا به برسی پایتون میپردازیم. دوره تخصصی «پایتون در علوم‌زمین» شامل کاربردهای واقعی و عملی می‌شود.
