## ۴.۱ مقدمه

تا اینجا برنامه‌هایی نوشتیم که مقادیر در خود کد تعریف شده بودند. اما در دنیای واقعی، برنامه‌ها باید بتوانند با کاربر **تعامل** داشته باشند — اطلاعات بگیرند و نتیجه نشان بدهند.

## ۴.۳ قالب‌بندی خروجی با f-string

f-string ساده‌ترین و خواناترین روش برای قالب‌بندی متن در پایتون است. کافی است قبل از رشته حرف `f` بگذارید و متغیرها را داخل `{}` قرار دهید.
</div>


```
product = "لپ‌تاپ"
price = 45000000
discount = 15

print(f"محصول: {product}")
print(f"قیمت: {price:,} تومان")
print(f"تخفیف: {discount}%")
```

<div dir="rtl" style="text-align:right;">

خروجی:

محصول: لپ‌تاپ
قیمت: 45,000,000 تومان
تخفیف: 15%

### قالب‌بندی اعداد

| فرمت | معنی | مثال | نتیجه |
|---|---|---|---|
| `:.2f` | دو رقم اعشار | `f"{3.14159:.2f}"` | `3.14` |
| `:.4f` | چهار رقم اعشار | `f"{3.14159:.4f}"` | `3.1416` |
| `:,` | جداکننده هزارگان | `f"{1000000:,}"` | `1,000,000` |
| `:>10` | تراز راست در ۱۰ کاراکتر | `f"{'سلام':>10}"` | `     سلام` |
| `:<10` | تراز چپ در ۱۰ کاراکتر | `f"{'سلام':<10}"` | `سلام     ` |
| `:^10` | تراز وسط در ۱۰ کاراکتر | `f"{'سلام':^10}"` | `  سلام   ` |
</div>


```
pi = 3.14159265
population = 1420000

print(f"عدد پی: {pi:.2f}")
print(f"عدد پی با دقت بیشتر: {pi:.6f}")
print(f"جمعیت: {population:,} نفر")
```

<div dir="rtl" style="text-align:right;">

خروجی:

عدد پی: 3.14
عدد پی با دقت بیشتر: 3.141593
جمعیت: 1,420,000 نفر

## ۴.۴ مثال کاربردی: فاکتور خرید
</div>


```
item1 = "کیبورد"
item2 = "ماوس"
item3 = "هدفون"

price1 = 850000
price2 = 320000
price3 = 1200000

total = price1 + price2 + price3
tax_rate = 0.09
tax = total * tax_rate
final_price = total + tax

print("=" * 35)
print(f"{'فاکتور خرید':^35}")
print("=" * 35)
print(f"{item1:<15} {price1:>15,} تومان")
print(f"{item2:<15} {price2:>15,} تومان")
print(f"{item3:<15} {price3:>15,} تومان")
print("-" * 35)
print(f"{'جمع کل':<15} {total:>15,} تومان")
print(f"{'مالیات (۹٪)':<15} {tax:>15,.0f} تومان")
print(f"{'مبلغ نهایی':<15} {final_price:>15,.0f} تومان")
print("=" * 35)
```

<div dir="rtl" style="text-align:right;">

خروجی:

===================================

فاکتور خرید            

===================================

کیبورد            850,000 تومان

ماوس              320,000 تومان

هدفون           1,200,000 تومان

-----------------------------------
جمع کل          2,370,000 تومان

مالیات (۹٪)       213,300 تومان

مبلغ نهایی      2,583,300 تومان

===================================

## ۴.۵ تابع input()

```
user_input = input("یک عدد وارد کنید: ")
print(type(user_input))    # <class 'str'>
```

## ۴.۶ تبدیل نوع داده (Type Conversion)

چون `input()` همیشه رشته برمی‌گرداند، برای انجام محاسبات باید نوع داده را تبدیل کنیم.

| تابع | تبدیل به | مثال |
|---|---|---|
| `int()` | عدد صحیح | `int("25")` → `25` |
| `float()` | عدد اعشاری | `float("3.14")` → `3.14` |
| `str()` | رشته | `str(100)` → `"100"` |
| `bool()` | بولین | `bool(0)` → `False` |
</div>


```
age_str = input("سن خود را وارد کنید: ")
age = int(age_str)
birth_year = 1404 - age
print(f"سال تولد شما: {birth_year}")
```

<div dir="rtl" style="text-align:right;">

اجرا:

سن خود را وارد کنید: 22

سال تولد شما: 1382

یا به صورت خلاصه‌تر:
</div>


```
age = int(input("سن خود را وارد کنید: "))
birth_year = 1404 - age
print(f"سال تولد شما: {birth_year}")
```

## ۴.۷ مثال کاربردی: ماشین حساب ساده
</div>


```
print("=== ماشین حساب ===")

num1 = float(input("عدد اول: "))
num2 = float(input("عدد دوم: "))

addition       = num1 + num2
subtraction    = num1 - num2
multiplication = num1 * num2

print("\n--- نتایج ---")
print(f"{num1} + {num2} = {addition}")
print(f"{num1} - {num2} = {subtraction}")
print(f"{num1} × {num2} = {multiplication}")

if num2 != 0:
    division = num1 / num2
    print(f"{num1} ÷ {num2} = {division:.4f}")
else:
    print("تقسیم بر صفر امکان‌پذیر نیست!")
```

<div dir="rtl" style="text-align:right;">

اجرا:

=== ماشین حساب ===

عدد اول: 15

عدد دوم: 4

--- نتایج ---

15.0 + 4.0 = 19.0

15.0 - 4.0 = 11.0

15.0 × 4.0 = 60.0

15.0 ÷ 4.0 = 3.7500

## ۴.۸ مثال کاربردی: محاسبه نمره نهایی
</div>


```
print("=== محاسبه نمره نهایی ===\n")

student_name = input("نام دانشجو: ")
midterm      = float(input("نمره میان‌ترم (از ۲۰): "))
final        = float(input("نمره پایان‌ترم (از ۲۰): "))
assignment   = float(input("نمره تکلیف (از ۲۰): "))

# میانگین وزنی: میان‌ترم ۳۰٪، پایان‌ترم ۵۰٪، تکلیف ۲۰٪
weighted_avg = (midterm * 0.30) + (final * 0.50) + (assignment * 0.20)

passed = weighted_avg >= 10

print(f"\n--- کارنامه {student_name} ---")
print(f"میان‌ترم:      {midterm:.1f}")
print(f"پایان‌ترم:     {final:.1f}")
print(f"تکلیف:        {assignment:.1f}")
print(f"نمره نهایی:   {weighted_avg:.2f}")
print(f"وضعیت:        {'قبول' if passed else 'مردود'}")
```

<div dir="rtl" style="text-align:right;">

اجرا:

=== محاسبه نمره نهایی ===

نام دانشجو: رضا احمدی

نمره میان‌ترم (از ۲۰): 14

نمره پایان‌ترم (از ۲۰): 16

نمره تکلیف (از ۲۰): 18

--- کارنامه رضا احمدی ---

میان‌ترم:      14.0

پایان‌ترم:     16.0

تکلیف:        18.0

نمره نهایی:   15.80

وضعیت:        قبول

## ۴.۱۰ تمرین‌های عملی

### تمرین ۱ — تبدیل دما

برنامه‌ای بنویسید که دما را به سلسیوس از کاربر بگیرد و به فارنهایت تبدیل کند:

F = (9/5) * C + 32

### تمرین ۲ — محاسبه مساحت

برنامه‌ای بنویسید که طول و عرض یک مستطیل را از کاربر بگیرد و مساحت و محیط آن را نمایش دهد.

### تمرین ۳ — فاکتور شخصی‌سازی‌شده

برنامه‌ای بنویسید که:
1. نام کاربر را بگیرد
2. سه محصول و قیمت آن‌ها را از کاربر بگیرد
3. یک فاکتور مرتب با جمع کل نمایش دهد

## ۴.۱۱ خلاصه

| تابع | کاربرد |
|---|---|
| `print()` | نمایش خروجی |
| `input()` | دریافت ورودی (همیشه str) |
| `int()` | تبدیل به عدد صحیح |
| `float()` | تبدیل به عدد اعشاری |
| `str()` | تبدیل به رشته |

نکات کلیدی:
- خروجی `input()` همیشه رشته است — قبل از محاسبه تبدیل کنید
- f-string ساده‌ترین روش قالب‌بندی متن است
- با `:.2f` می‌توان تعداد ارقام اعشار را کنترل کرد
- با `:,` می‌توان اعداد بزرگ را خواناتر نمایش داد
